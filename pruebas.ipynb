```python
from dotenv import load_dotenv
import openai
import os
import json
from datetime import datetime

load_dotenv()

client = openai.OpenAI(
    api_key=os.getenv("POLIGPT_API_KEY"),
    base_url=os.getenv("POLIGPT_URL")
)
models = ["gpt-oss-120b"]
for model in models:
    print("-" * 50)
    print(f"Testing model: {model}")
    ts = datetime.now().isoformat()
    try:
        response = client.chat.completions.create(
            model=model,
            messages = [
                {
                    "role": "user",
                    "content": "Cuéntame todo lo que sepas sobre cómo ganar jugando al catan"
                }
            ]
        )
        te = datetime.now().isoformat()
        message = response.choices[0].message.content
        respmodel = response.model
        time = datetime.fromisoformat(te) - datetime.fromisoformat(ts)
        print(f"Model: {respmodel}\nMessage:\n{message}\nTime: {time}")
    except Exception as e:
        print(f"Model: {model}\nError: {str(e)}\n")
```
